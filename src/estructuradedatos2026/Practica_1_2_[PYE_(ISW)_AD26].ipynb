### Problema 1: Tiempo de Secado

1.	Se registran las siguientes mediciones para el tiempo de secado (en horas) de cierta marca de pintura esmaltada:

- 3.4, 2.5, 4.8, 2.9, 3.6, 2.8, 3.3, 5.6, 3.7, 2.8, 4.4, 4.0, 5.2, 3.0, 4.8

Suponga que las mediciones constituyen una muestra aleatoria simple. Calcule, para la muestra:
1. **Medidas de Tendencia Central** (Media, Mediana, Moda)
2. **Media Recortada al 10%**
3. **Medidas de Variabilidad** (Rango, Desviación Estándar, Varianza)
4. **Cuartiles**
5. **Coeficientes de Asimetría y Curtosis**
6. **Brinde una interpretación general del problema**

```python
# Aquí se debe ingresar el código para importar librerías, definir listas y comandos para realizar los cálculos correspondientes
import numpy as np
from scipy import stats
from scipy.stats import trim_mean

#Mediciones del tiempo secado
tiempo_secado = [3.4, 2.5, 4.8, 2.9, 3.6, 2.8, 3.3, 5.6, 3.7, 2.8, 4.4, 4.0, 5.2, 3.0, 4.8]

```

**Parte 1:** Calcule las Medidas de Tendencia Central (Media, Mediana, Moda) para los tiempos de secado.

```python
# Ingresar código para calcular las medidas de tendencia central
media = np.mean(tiempo_secado)
mediana = np.median(tiempo_secado)
moda = stats.mode(tiempo_secado)[0]

print("Medidas de Tendencia Central:\n")
print(f"Media: {media:.2f}")
print(f"Mediana: {mediana:.2f}")
print(f"Moda: {moda:.2f}")
```

**Parte 2:** Calcule la Media Recortada al 20% para los tiempos de secado.

```python
# Ingresar código para calcular la media recortada al 10%, en el problema dice recortada al 10%
media_recortada_10 = trim_mean(tiempo_secado, proportiontocut=0.1)

print("Media recortada:\n")
print(f"Media recortada 10: {media_recortada_10:.2f}")
```

```python
# Ingresar código para calcular la media recortada al 20%
media_recortada_20 = trim_mean(tiempo_secado, proportiontocut=0.2)

print("Media recortada:\n")
print(f"Media recortada 20: {media_recortada_10:.2f}")
```

**Parte 3:** Calcule las Medidas de Variabilidad (Rango, Desviación Estándar, Varianza) para los tiempos de secado.

```python
# Ingresar código para calcular las medidas de variabilidad
rango = np.ptp(tiempo_secado)
desviación_estandar = np.std(tiempo_secado, ddof=1)
varianza = np.var(tiempo_secado, ddof=1)

print("Medidades de Variabilidad: \n")
print(f"Rango: {rango:.2f}")
print(f"Desviación Estándar: {desviación_estandar:.2f}")
print(f"Varianza: {varianza:.2f}")
```

**Parte 4:** Calcule los Cuartiles para los tiempos de secado.

```python
# Ingresar código para calcular los cuartiles
q1 = np.quantile(tiempo_secado, .25)
q2 = np.quantile(tiempo_secado, .50)
q3 = np.quantile(tiempo_secado, .75)

print("Cuartiles:\n")
print(f"Q1 (25%): {q1:.2f}")
print(f"Q2 (50%): {q2:.2f}")
print(f"Q3 (75%): {q3:.2f}")
```

**Parte 5:** Calcule el Coeficiente de Asimetría y la Curtosis para los tiempos de secado.

```python
# Ingresar código para calcular los coeficientes de asimetría y curtosis
asimetria = stats.skew(tiempo_secado, bias=False)
curtosis = stats.kurtosis(tiempo_secado, bias=False)

print("Asimetría Y Curtosis:\n")
print(f"Asimetría: {asimetria:.2f}")
print(f"Curtosis: {curtosis:.2f}")
```

**Parte 6:** Brinde una interpretación general del problema

Los datos muestran una ligera asimetría que indica que algunos tiempos de secado son mayores que el promedio.

### Problema 2: Efecto de un Régimen de Salud

Veinte hombres adultos de entre 30 y 40 años de edad participaron en un estudio para evaluar el efecto de cierto régimen de salud, que incluye dieta y ejercicio, en el colesterol sanguíneo. Se eligieron aleatoriamente diez para el grupo de control y los otros diez se asignaron para participar en el régimen como el grupo de tratamiento durante un periodo de seis meses. Los siguientes datos muestran la reducción en el colesterol que experimentaron en ese periodo los 20 sujetos:

- **Grupo de Control:** 7, 3, -4, 14, 2, 5, 22, -7, 9, 5
- **Grupo de Tratamiento:** -6, 5, 9, 4, 4, 12, 37, 5, 3, 3

Calcule, para cada muestra:
1. **Medidas de Tendencia Central** (Media, Mediana, Moda)
2. **Media Recortada al 10%**
3. **Medidas de Variabilidad** (Rango, Desviación Estándar, Varianza)
4. **Cuartiles**
5. **Coeficientes de Asimetría y Curtosis**
6. **Brinde una interpretación general del problema**

```python
# Aquí se debe ingresar el código para importar librerías, definir listas y comandos para realizar los cálculos correspondientes
import numpy as np
from scipy import stats
from scipy.stats import trim_mean

#Datos
gpo_control = [7, 3, -4, 14, 2, 5, 22, -7, 9, 5]
gpo_tratamiento = [-6, 5, 9, 4, 4, 12, 37, 5, 3, 3]
```

**Parte 1:** Calcule las Medidas de Tendencia Central (Media, Mediana, Moda) para los datos de ambas compañías.

```python
# Ingresar código para calcular las medidas de tendencia central de ambas compañías
def calcular_estadisticas(datos):
  media = np.mean(datos)
  mediana = np.median(datos)
  moda = stats.mode(datos)[0]
  return media, mediana, moda

# Medidas de Tendencia Central
media_control, mediana_control, moda_control = calcular_estadisticas(gpo_control)
media_tratamiento, mediana_tratamiento, moda_tratamiento = calcular_estadisticas(gpo_tratamiento)

print("Medidas de Tendencia Central en Grupo de Control:")
print(f"Media: {media_control:.2f}")
print(f"Mediana: {mediana_control:.2f}")
print(f"Moda: {moda_control:.2f}\n")
print("Medidas de Tendencia Central en Grupo de Tratamiento:")
print(f"Media: {media_tratamiento:.2f}")
print(f"Mediana: {mediana_tratamiento:.2f}")
print(f"Moda: {moda_tratamiento:.2f}")
```

**Parte 2:** Calcule la Media Recortada al 10% para los datos de ambas compañías.

```python
# Ingresar código para calcular la media recortada al 10% de ambas compañías
media_recortada_control = trim_mean(gpo_control, proportiontocut=0.1)
media_recortada_tratamiento = trim_mean(gpo_tratamiento, proportiontocut=0.1)

print("Media recortada Grupo de Control:")
print(f"Media Recortada al 10%: {media_recortada_control:.2f}\n")
print("Media recortada Grupo de Tratamiento:")
print(f"Media Recortada al 10%: {media_recortada_tratamiento:.2f}")
```

**Parte 3:** Calcule las Medidas de Variabilidad (Rango, Desviación Estándar, Varianza) para los datos de ambas compañías

```python
# Ingresar código para calcular las medidas de variabilidad de ambas compañías
def calcular_variabilidad(datos):
  rango = np.ptp(datos)
  desviación_estandar = np.std(datos, ddof=1)
  varianza = np.var(datos, ddof=1)
  return rango, desviación_estandar, varianza

rango_control, desviacion_estandar_control, varianza_control = calcular_variabilidad(gpo_control)
rango_tratamiento, desviacion_estandar_tratamiento, varianza_tratamiento = calcular_variabilidad(gpo_tratamiento)

print("\nMedidas de Variabilidad Grupo de Control:")
print(f"Rango: {rango_control:.2f}")
print(f"Desviación Estándar: {desviacion_estandar_control:.2f}")
print(f"Varianza: {varianza_control:.2f}\n")
print("Medidas de Variabilidad Grupo de Tratamiento:")
print(f"Rango: {rango_tratamiento:.2f}")
print(f"Desviación Estándar: {desviacion_estandar_tratamiento:.2f}")
print(f"Varianza: {varianza_tratamiento:.2f}")
```

**Parte 4:** Calcule los Cuartiles para los datos de ambas compañías

```python
# Ingresar código para calcular los cuartiles de ambas compañías
def calcular_cuartiles(datos):
  q1 = np.quantile(datos, .25)
  q2 = np.quantile(datos, .50)
  q3 = np.quantile(datos, .75)
  return q1, q2, q3

q1_control, q2_control, q3_control = calcular_cuartiles(gpo_control)
q1_tratamiento, q2_tratamiento, q3_tratamiento = calcular_cuartiles(gpo_tratamiento)

print("Cuartiles Grupo de Control:")
print(f"Q1 (25%): {q1_control:.2f}")
print(f"Q2 (50%): {q2_control:.2f}")
print(f"Q3 (75%): {q3_control:.2f}\n")
print("Cuartiles Grupo de Tratamiento:")
print(f"Q1 (25%): {q1_tratamiento:.2f}")
print(f"Q2 (50%): {q2_tratamiento:.2f}")
print(f"Q3 (75%): {q3_tratamiento:.2f}")
```

**Parte 5:** Calcule el Coeficiente de Asimetría y la Curtosis para los datos de ambas compañías

```python
# Ingresar código para calcular los coeficientes de asimetría y curtosis de ambas compañías

asimetria_gt = stats.skew(gpo_tratamiento, bias=False)
curtosis_gt = stats.kurtosis(gpo_tratamiento, bias=False)
asimetria_gc = stats.skew(gpo_control, bias=False)
curtosis_gc = stats.kurtosis(gpo_control, bias=False)

print("Asimetría Y Curtosis Grupo de Control:")
print(f"Asimetría: {asimetria_gc:.2f}")
print(f"Curtosis: {curtosis_gc:.2f}\n")
print("Asimetría Y Curtosis Grupo de Tratamiento:")
print(f"Asimetría: {asimetria_gt:.2f}")
print(f"Curtosis: {curtosis_gt:.2f}")
```

**Parte 6:** Brinde una interpretación general del problema

El grupo de tratamiento muestra un mayor promedio en reducción de colesterol que el grupo de control, pero también tuvo una mayor variación y asimetría debido a que unas muestras tenían valores más elevados. Por lo tanto, el grupo de tratamiento presentó una mayor reducción promedio, pero con resultados menos uniformes entre las muestras.
