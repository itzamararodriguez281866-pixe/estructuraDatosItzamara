### Problema 1: Vida de las Baterías para Automóvil

Considérense los siguientes datos que especifican la "vida" de 40 baterías para automóvil similares, registradas al décimo de año más cercano. Las baterías se garantizan por tres años.

        2.2, 4.1, 3.5, 4.5, 3.2, 3.7, 3.0, 2.6,
        3.4, 1.6, 3.1, 3.3, 3.8, 3.1, 4.7, 3.7,
        2.5, 4.3, 3.4, 3.6, 2.9, 3.3, 3.9, 3.1,
        3.3, 3.1, 3.7, 4.4, 3.2, 4.1, 1.9, 3.4,
        4.7, 3.8, 3.2, 2.6, 3.9, 3.0, 4.2, 3.5

Realice, para los datos de la muestra:
1. **Diagrama de tallo y hojas**
2. **Distribución de Frecuencias**
3. **Histograma**

```python
# Ingresar el código para importar librerías y definir listas
import numpy as np
import math
import matplotlib.pyplot as plt

datos = [ 2.2, 4.1, 3.5, 4.5, 3.2, 3.7, 3.0, 2.6,
          3.4, 1.6, 3.1, 3.3, 3.8, 3.1, 4.7, 3.7,
          2.5, 4.3, 3.4, 3.6, 2.9, 3.3, 3.9, 3.1,
          3.3, 3.1, 3.7, 4.4, 3.2, 4.1, 1.9, 3.4,
          4.7, 3.8, 3.2, 2.6, 3.9, 3.0, 4.2, 3.5]

datos = np.array(datos)
```

**Parte 1:** Diagrama de Tallo y Hojas

```python
# Ingresar código para realizar el Diagrama de Tallo y Hojas

datos_ordenados = sorted(datos)

#Diccionario vacio
tallo_hojas = {}

#Separar tallo y hojas
for valor in datos_ordenados:
  tallo, hoja = str(valor).split(".")

  tallo_hojas.setdefault(tallo, [])
  tallo_hojas[tallo].append(hoja)

print("Diagrama Tallo y Hojas")
for tallo, hojas in tallo_hojas.items():
  hojas = " ".join(hojas)
  print(tallo, "|", hojas)
```

**Parte 2:** Distribución de Frecuencias

```python
# Ingresar código para realizar la Distribución de Frecuencias
# Opción 1: Especificar número de intervalos
n = len(datos)

numero_clases = math.ceil(math.sqrt(n))
print("Intervalos:", numero_clases)

#Obetener el valor minimo y  el vlor maximo
minimo = min(datos)
maximo = max(datos)

#ncho de la clase
ancho_clase = math.ceil(((maximo - minimo)/ numero_clases) * 10 )/10
print("Ancho de clase: ", ancho_clase)

frecuencias, limites = np.histogram(datos, bins=numero_clases)

print("Frecuencias:", frecuencias)
print("Límites:", limites)


```

```python
# Ingresar código para realizar la Distribución de Frecuencias
# Opción 2: Especificar los límites de cada clase
limite_inicial = minimo - math.floor(ancho_clase * 5) / 10

limites = np.linspace(limite_inicial, limite_inicial + numero_clases * ancho_clase, numero_clases + 1)

limites = [round(x, 1) for x in limites]

frecuencias, limites = np.histogram(datos, bins=limites)

print("Límites:", limites)
print("Frecuencias:", frecuencias)
```

**Parte 3:** Realice el Histograma

```python
# Ingresar código para realizar el Histograma
# Opción 1: Especificar número de intervalos

plt.hist(datos, bins= numero_clases, edgecolor= "black", color= "purple")
plt.title("Vida de las baterias para automovil")
plt.xlabel("Vida de la bateria (años)")
plt.ylabel("Frecuencia")

plt.show()

```

```python
# Opción 2: Especificar los límites de cada clase
plt.hist(datos, bins= limites, edgecolor= "black", color= "purple")

plt.xticks(limites)
plt.xlabel("Vida de la bateria (años)")
plt.ylabel("Frecuencia")
```
