### Problema 1: Absorción de Agua en Fibra de Algodón

Según la revista *Chemical Engineering*, una propiedad importante de una fibra es su absorción del agua. Se toma una muestra aleatoria de 20 pedazos de fibra de algodón y se mide la absorción de cada uno. Los valores de absorción son los siguientes:
- 18.71, 21.41, 20.72, 21.81, 19.29, 22.43, 20.17, 23.71, 19.44, 20.50
- 18.92, 20.33, 23.00, 22.85, 19.25, 21.77, 22.11, 19.77, 18.04, 21.12

Calcule, para los datos de la muestra:
1. **Medidas de Tendencia Central** (Media, Mediana, Moda)
2. **Media Recortada al 20%**
3. **Medidas de Variabilidad** (Rango, Desviación Estándar, Varianza)

```python
# Aquí se debe ingresar el código para importar librerías, definir listas y comandos para realizar los cálculos correspondientes

# Importación de librerías
import numpy as np
from scipy import stats
from scipy.stats import trim_mean

# Datos de absorción de fibra
absorcion_fibra = [
    18.71, 21.41, 20.72, 21.81, 19.29, 22.43, 20.17, 23.71, 19.44, 20.50,
    18.92, 20.33, 23.00, 22.85, 19.25, 21.77, 22.11, 19.77, 18.04, 22.11
]
```

**Parte 1:** Calcule las Medidas de Tendencia Central (Media, Mediana, Moda) para los datos de absorción de agua.

```python
# Medidas de Tendencia Central
media = np.mean(absorcion_fibra)
mediana = np.median(absorcion_fibra)
moda = stats.mode(absorcion_fibra)[0]

print("\033[1m"+"Medidas de Tendencia Central:"+"\033[0m")
print(f"Media: {media:.2f}")
print(f"Mediana: {mediana:.2f}")
print(f"Moda: {moda:.2f}")
```

**Parte 2:** Calcule la Media Recortada al 20% para los datos de absorción de agua.

```python
# Media Recortada al 20%
media_recortada_20 = trim_mean(absorcion_fibra, proportiontocut=0.2)

print("\033[1m"+"\nMedia Recortada:"+"\033[0m")
print(f"Media recortada al 20%: {media_recortada_20:.2f}")
```

**Parte 3:** Calcule las Medidas de Variabilidad (Rango, Desviación Estándar, Varianza) para los datos de absorción de agua.

```python
# Medidas de Variabilidad
rango = np.ptp(absorcion_fibra)  # Rango
desviacion_estandar = np.std(absorcion_fibra, ddof=1)  # Desviación Estándar
varianza = np.var(absorcion_fibra, ddof=1)  # Varianza

print("\033[1m"+"\nMedidas de Variabilidad:"+"\033[0m")
print(f"Rango: {rango:.2f}")
print(f"Desviación Estándar: {desviacion_estandar:.2f}")
print(f"Varianza: {varianza:.2f}")
```

**Parte 4:** Calcule los Cuartiles para los datos de absorción de agua

```python
# Cuartiles
q1 = np.quantile(absorcion_fibra, .25)
q2 = np.quantile(absorcion_fibra, .50)
q3 = np.quantile(absorcion_fibra, .75)

print("\033[1m\nCuartiles:\033[0m")
print(f"Q1 (25%): {q1:.2f}")
print(f"Q2 (50% o Mediana): {q2:.2f}")
print(f"Q3 (75%): {q3:.2f}")
```

**Parte 5:** Calcule el Coeficiente de Asimetría para los datos de absorción de agua

```python
# Coeficiente de Asimetría
# bias=False asegura que sea el cálculo muestral
asimetria = stats.skew(absorcion_fibra, bias=False)

print("\033[1m\nCoeficiente de Asimetría:\033[0m")
print(f"Asimetría: {asimetria:.4f}")
```

**Parte 6:** Calcule la Curtosis para los datos de absorción de agua

```python
# Curtosis
# bias=False asegura que sea el cálculo muestral de Fisher
curtosis = stats.kurtosis(absorcion_fibra, bias=False)

print("\033[1m\nCurtosis:\033[0m")
print(f"Curtosis: {curtosis:.4f}")
```

**Parte 7:** Brinde una interpretación general del problema

Las fibras de algodón presentan un promedio de absorción de 20.82. Los datos no muestran una variación drástica, por lo que se puede determinar que presentan poca asimetría.

### Problema 2: Comparación de Flexibilidad en Varillas de Acero

En un estudio realizado por el Departamento de Ingeniería Mecánica del Tecnológico de Virginia se compararon las varillas de acero que abastecen dos compañías diferentes. Se fabricaron diez resortes de muestra con las varillas de metal proporcionadas por cada una de las compañías y se registraron sus medidas de flexibilidad. Los datos son los siguientes:

- **Compañía A:** 9.3, 8.8, 6.8, 8.7, 8.5, 6.7, 8.0, 6.5, 9.2, 7.0
- **Compañía B:** 11.0, 9.8, 9.9, 10.2, 10.1, 9.7, 11.0, 11.1, 10.2, 9.6

Calcule, para cada muestra:
1. **Medidas de Tendencia Central** (Media, Mediana, Moda)
2. **Media Recortada al 10%**
3. **Medidas de Variabilidad** (Rango, Desviación Estándar, Varianza)


```python
# Aquí se debe ingresar el código para importar librerías, definir listas y comandos para realizar los cálculos correspondientes

import numpy as np
from scipy import stats
from scipy.stats import trim_mean

# Datos de flexibilidad
compania_A = [9.3, 8.8, 6.8, 8.7, 8.5, 6.7, 8.0, 6.5, 9.3, 7.0]
compania_B = [11.0, 9.8, 9.9, 10.2, 10.1, 9.7, 11.0, 11.1, 10.2, 9.6]
```

**Parte 1:** Calcule las Medidas de Tendencia Central (Media, Mediana, Moda) para los datos de ambas compañías.

```python
# Función para calcular las estadísticas
def calcular_estadisticas(datos):
    media = np.mean(datos)
    mediana = np.median(datos)
    moda = stats.mode(datos)[0]
    return media, mediana, moda

# Medidas de Tendencia Central
media_A, mediana_A, moda_A = calcular_estadisticas(compania_A)
media_B, mediana_B, moda_B = calcular_estadisticas(compania_B)

print("\033[1m"+"Medidas de Tendencia Central Compañía A:"+"\033[0m")
print(f"Media: {media_A:.2f}")
print(f"Mediana: {mediana_A:.2f}")
print(f"Moda: {moda_A:.2f}")
print("\033[1m"+"Medidas de Tendencia Central Compañía B:"+"\033[0m")
print(f"Media: {media_B:.2f}")
print(f"Mediana: {mediana_B:.2f}")
print(f"Moda: {moda_B:.2f}")
```

**Parte 2:** Calcule la Media Recortada al 10% para los datos de ambas compañías.

```python
# Media Recortada al 10%
media_recortada_A = trim_mean(compania_A, proportiontocut=0.1)
media_recortada_B = trim_mean(compania_B, proportiontocut=0.1)

print("\033[1m"+"\nMedia Recortada Compañía A:"+"\033[0m")
print(f"Media Recortada al 10%: {media_recortada_A:.2f}")
print("\033[1m"+"\nMedia Recortada Compañía B:"+"\033[0m")
print(f"Media Recortada al 10%: {media_recortada_B:.2f}")
```

**Parte 3:** Calcule las Medidas de Variabilidad (Rango, Desviación Estándar, Varianza) para los datos de ambas compañías

```python
def calcular_variabilidad(datos):
    rango = np.ptp(datos)  # Rango
    desviacion_estandar = np.std(datos, ddof=1)  # Desviación Estándar
    varianza = np.var(datos, ddof=1)  # Varianza
    return rango, desviacion_estandar, varianza

# Medidas de Variabilidad
rango_A, desviacion_estandar_A, varianza_A = calcular_variabilidad(compania_A)
rango_B, desviacion_estandar_B, varianza_B = calcular_variabilidad(compania_B)

print("\033[1m"+"\nMedidas de Variabilidad Compañía A:"+"\033[0m")
print(f"Rango: {rango_A:.2f}")
print(f"Desviación Estándar: {desviacion_estandar_A:.2f}")
print(f"Varianza: {varianza_A:.2f}")
print("\033[1m"+"\nMedidas de Variabilidad Compañía B:"+"\033[0m")
print(f"Rango: {rango_B:.2f}")
print(f"Desviación Estándar: {desviacion_estandar_B:.2f}")
print(f"Varianza: {varianza_B:.2f}")
```

**Parte 4**: Calcule los Cuartiles para los datos de ambas compañías

```python
# Cuartiles
def calcular_cuartiles(datos):
    q1 = np.percentile(datos, 25)
    q2 = np.percentile(datos, 50)
    q3 = np.percentile(datos, 75)
    return q1, q2, q3

q1_A, q2_A, q3_A = calcular_cuartiles(compania_A)
q1_B, q2_B, q3_B = calcular_cuartiles(compania_B)

print("\033[1m\nCuartiles Compañía A:\033[0m")
print(f"Q1: {q1_A:.2f} | Q2: {q2_A:.2f} | Q3: {q3_A:.2f}")
print("\033[1m\nCuartiles Compañía B:\033[0m")
print(f"Q1: {q1_B:.2f} | Q2: {q2_B:.2f} | Q3: {q3_B:.2f}")
```

**Parte 5:** Calcule el Coeficiente de Asimetría para los datos de ambas compañías

```python
# Asimetría
asimetria_A = stats.skew(compania_A, bias=False)
asimetria_B = stats.skew(compania_B, bias=False)

print("\033[1m\nAsimetría Compañía A:\033[0m", f"{asimetria_A:.4f}")
print("\033[1mAsimetría Compañía B:\033[0m", f"{asimetria_B:.4f}")
```

**Parte 6:** Calcule la Curtosis para los datos de ambas compañías

```python
# Curtosis
curtosis_A = stats.kurtosis(compania_A, bias=False)
curtosis_B = stats.kurtosis(compania_B, bias=False)

print("\033[1m\nCurtosis Compañía A:\033[0m", f"{curtosis_A:.4f}")
print("\033[1mCurtosis Compañía B:\033[0m", f"{curtosis_B:.4f}")
```

**Parte 7:** Brinde una interpretación general del problema

La compañía B muestra una flexibilidad promedio mayor que compañía A. Presentando una menor variación, con una desviación estándar de 0.57 contra 1.11 de la compañía A. La compañía B tiene valores de flexibilidad más altos y estables.
